# geohash-kit demo
Python API walk-through on the Central Park outline in `examples/central_park.geojson`.
Run from the repository root (or Colab after `pip install git+https://github.com/AmmarYasser455/geohash-kit`).

```python
import os, urllib.request
import geohash_kit as gk
PATH = 'examples/central_park.geojson'
if not os.path.exists(PATH):
    os.makedirs('examples', exist_ok=True)
    urllib.request.urlretrieve('https://raw.githubusercontent.com/AmmarYasser455/geohash-kit/main/' + PATH, PATH)
gh = gk.encode(57.64911, 10.40744, 11)
print(gh)
```

```python
lat, lon = gk.decode('u4pruydqqvj')
print(round(lat, 5), round(lon, 5))
print(gk.neighbours('u4pruyd'))
```

```python
fc = gk.cover_geojson(PATH, 6)
cells = [f['properties']['geohash'] for f in fc['features']]
print(len(cells), cells)
assert gk.encode(40.7829, -73.9654, 6) in cells  # Central Park's centre
print(sum(f['properties']['area_km2'] for f in fc['features']))
```
